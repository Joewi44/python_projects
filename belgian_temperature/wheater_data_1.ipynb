```python
import numpy as np

with open("data_files/belgian_temp.npy", "rb") as fh:
    df_arr = np.load(fh)
    year_range_arr = np.load(fh)
    station_name_arr = np.load(fh)
    files_arr = np.load(fh)
```

1. Where and when was the highest temperature observed and the lowest?
2. Which station had the coldest summer temperature (average over the months June, July and August) and in which year was that?
3. Which were on average the warmest and coldest year in Belgium?
4. When was the biggest temperature difference observed between consecutive months in Belgium?

```python
# 1.
max_temp_all = np.nanmax(df_arr)
station_idx, year_idx, month_idx = np.where(df_arr == max_temp_all)
print(
    f"Max temp is {max_temp_all} measured in station {station_name_arr[station_idx][0]} "
    f"year: {year_range_arr[year_idx][0]} and month {month_idx + 1} \n"
    f"file: {files_arr[station_idx]}\n"
    )

min_temp_all = np.nanmin(df_arr)
station_idx, year_idx, month_idx = np.where(df_arr == min_temp_all)
print(
    f"Min temp is {min_temp_all} measured in station {station_name_arr[station_idx][0]} "
    f"year: {year_range_arr[year_idx[0]]} and month {month_idx +1 } \n"
    f"file: {files_arr[station_idx]}"
    )
```

```python
#2. Which station had the coldest summer temperature (average over the months June, July and August) and in which year was that?
df_avg_summer = np.nanmean(df_arr[:,:,5:8], axis=2) # SHAPE (16, 145)
coldest_summer = np.nanmin(df_avg_summer)
station_idx, year_idx = np.where(df_avg_summer == coldest_summer)
print(f"Coldest summer average was for {station_name_arr[station_idx]} in {year_range_arr[year_idx]} "
      f"with temp of {coldest_summer:.2f}C°\n"
      f"file {files_arr[station_idx]}")
```

```python
#2. Which station had the coldest summer temperature (average over the months June, July and August) and in which year was that?
df_arr_summer = df_arr[:,:,5:8]         #SHAPE (16, 145, 3)
#print(station_name_arr[0], files_arr[0])
coldest_summer = np.nanmin(df_arr_summer)
station_idx, year_idx, month_idx = np.where(df_arr_summer == coldest_summer)
print(f"Coldest summer was for {station_name_arr[station_idx]} in {year_range_arr[year_idx]} {month_idx + 6} "
      f"with temp of {coldest_summer}C°\n"
      f"file {files_arr[station_idx]}")
```

```python
# 3. Which were on average the warmest and coldest year in Belgium?

average_temp = np.nanmean(df_arr, axis=2) # shape (16, 145)
min_average_temp = np.nanmin(average_temp)
station_idx, year_idx = np.where(average_temp == min_average_temp)
print(f"The average coldest year was measured in {station_name_arr[station_idx]} in year {year_range_arr[year_idx]}"
      f" with temp of {min_average_temp}C°\n"
      f"file {files_arr[station_idx]}")

max_average_temp = np.nanmax(average_temp)
station_idx, year_idx = np.where(average_temp == max_average_temp)
print(f"The average warmest year was measured in {station_name_arr[station_idx]} in year {year_range_arr[year_idx]}"
      f" with temp of {max_average_temp}C°\n"
      f"file {files_arr[station_idx]}")

#Verify result
print(f"Verify average is {max_average_temp == np.sum(df_arr[station_idx, year_idx, :])/12}")
```

```python
# 4. When was the biggest temperature difference observed between consecutive months in Belgium?

df_difference = np.abs(np.diff(df_arr, axis=2)) #(16, 145, 11)
max_diff = np.nanmax(df_difference)
station_idx, year_idx, month_idx = np.where(df_difference == max_diff)
print(f"Station: {station_name_arr[station_idx][0]} year: {year_range_arr[year_idx]} "
      f"month: {month_idx+1} max difference: {max_diff:.2f}°C")
```
