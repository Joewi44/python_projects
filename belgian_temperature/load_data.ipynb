# Belgian Temperature project with numpy

I will make a 3D array to store all the readings.
Not every station start at same year

```python
import numpy as np
import os
```

## Store station name into array

```python
working_dir = "/Users/uwev/PycharmProjects/data-science/data_files/belgian_temperature/"
files_arr = np.array([os.listdir(working_dir)], dtype=str).reshape(16)  #SHAPE IS (16,)
station_name_arr = np.empty((len(files_arr),), dtype='U30')             #SHAPE IS (16,)

for i, file in enumerate(files_arr):
    with open(os.path.join(working_dir,file), 'r') as fh:
        station_name_arr[i] = fh.readlines()[3].replace("# Series name : ", "").strip()

print(files_arr)
station_name_arr
```

## get year range

```python
year_range_arr = np.array([], dtype=int)
for file in files_arr:
    df_year = np.genfromtxt(os.path.join(working_dir,file), delimiter=',', usecols=0, dtype=int, skip_header=9)
    year_range_arr = np.union1d(year_range_arr, df_year)

year_range_arr
```

```python
months = 12
df_arr = np.full(shape=(len(station_name_arr), len(year_range_arr)+1, months), fill_value=np.nan)
for station_idx, file in enumerate(files_arr):
    df = np.genfromtxt(os.path.join(working_dir,file), delimiter=',', skip_header=9)
    for row in df:
        temp = int(row[0])
        diff_temp = temp - np.min(year_range_arr)
        df_arr[station_idx, diff_temp] = row[1:13]
# chack for one station the last 5 rows & the firts 10 rows
print(df_arr[0,-6:-1,:]) # LAST VALUES CORRESPOND with "ho_monthly_series_203_tn_v2024_01.csv"
print(df_arr[0,0:10,:]) # start at 1889 so 9 NAN LINES
```

```python
print(df_arr[6,-5:-1,:]) #  "ho_monthly_series_208_tn_v2024_01.csv" 1880 - 2023 SO NO NAN VALLUES
print(df_arr[6,0:10,:])
```

```python
#Store to NPY file so I can do calculations in other python file
with open("data_files/belgian_temp.npy", "wb") as fw:
    np.save(fw, df_arr)
    np.save(fw, year_range_arr)
    np.save(fw, station_name_arr)
    np.save(fw, files_arr)
```
