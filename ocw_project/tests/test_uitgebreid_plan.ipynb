### **Doel van onderhoudsmaatregelen in de OCW-systematiek**

De onderhoudsmaatregelen worden gebruikt om verschillende onderhoudsstrategieën te vergelijken, niet als exacte uitvoeringsplannen. Elk project vereist specifieke voorbereiding en beoordeling.

---

### **9.1 Soorten onderhoudsmaatregelen**

De OCW-systematiek onderscheidt drie hoofdgroepen:

1. **Lokale reparaties**: Klein, plaatselijk herstel aan de verharding.
2. **Algemene reparaties**: Volledige behandeling van het oppervlak van een wegvak, zonder ingreep in de fundering.
3. **Versterking**: Structurele ingrepen ter verhoging van de draagkracht, inclusief reconstructies.

Een onderhoudsstrategie verloopt vaak als volgt:

* Eerst lokale reparaties,
* Vervolgens algemene reparaties,
* En uiteindelijk versterking naarmate de weg ouder wordt.

#### **9.1.1 Asfaltverharding**

* **Lokale reparaties**: o.a. vullen van scheuren, herstellen van kippennesten; gaan gemiddeld 8 jaar mee.
* **Algemene reparaties**:

  * Eenlaagse of tweelaagse bestrijking, slemafdichting.
  * Theoretisch 12–16 jaar levensduur, maar in praktijk vaak 5–10 jaar.
  * Andere varianten: affrezen, scheurremmende lagen, enz.
* **Versterking**:

  * Dunne of dikke overlaging (6–12 cm), of volledige reconstructie.
  * Levensduur theoretisch 17–30 jaar.
  * Alternatieven: gedeeltelijke vervanging, in-situ recycling.

#### **9.1.2 Betonverharding**

* **Lokale reparaties**: scheuren dichten, voegvulling; levensduur ±16 jaar.
* **Algemene reparatie**: vervangen van beschadigde betonplaten; levensduur ±30 jaar.
* **Versterking**: overlaging, vervanging of reconstructie; levensduur 38–42 jaar.

#### **9.1.3 Elementenverhardingen (straatstenen)**

* **Lokale reparaties**: herstraten.
* **Algemene reparatie**: volledig herstraten, inclusief straatlaag.
* **Versterking**: volledige reconstructie, inclusief fundering.

#### **9.1.4 Overwegingen**

De exacte technieken zijn minder belangrijk dan hun verhouding in kosten en effect. De onderhoudsmaatregel beïnvloedt visuele en structurele indexwaarden. Evolutiewetten voor de wegkwaliteit zijn afgeleid van de gekozen onderhoudscategorie.

---

### **9.2 Bepaling van eenheidsprijzen**

* Prijzen worden gebaseerd op gemiddelden uit recente projecten.
* Voor lokale reparaties: totale jaarlijkse uitgaven gedeeld door behandelde oppervlakte.
* Voor algemene reparaties en versterking: gemiddelden per type verharding.
* Extra investeringen (zoals fietspaden) kunnen al dan niet worden meegeteld, afhankelijk van de keuze van de wegbeheerder.

---

### **9.3 Economisch “spaarmodel”**

Om het optimale moment voor grotere onderhoudsmaatregelen te bepalen, wordt een economisch model gebruikt gebaseerd op:

* **Sparen** voor een toekomstige ingreep (algemene reparatie of versterking).
* **Jaarlijkse lokale reparaties**, die toenemen naarmate de weg ouder wordt.
* **Beslissingsregel**: Het economisch optimale tijdstip is wanneer de som van jaarlijkse spaarlast + lokale reparaties het laagst is.

Het model houdt rekening met:

* Indexering van kosten (i)
* Disconteringsvoet (r)
* Groei van lokale kosten (σ)

#### Bereken T_vereenvoudigd

- T = sqrt (B/A)
- A = de jaarlijkse investering in lokale reparaties.
- B = de investering in de algemene onderhoudsmaatregel.

Deze eenvoudige formule geeft meteen aan dat het vooral de verhouding tussen prijs B voor de op tijd-stip T uit te voeren algemene onderhoudsmaatregel en prijs A voor lokale reparaties is die het optimale tijdstip voor de uitvoering bepaalt, en dus niet om de absolute prijzen zelf.
De drempelwaarden die in de OCW-systematiek worden gebruikt, blijven dus geldig zolang de verhoudingen tussen de prijzen ongeveer gelijk blijven

```python
import numpy as np
import pandas as pd
from ocw_project.Maintenance_Manager import MaintenanceManager



def _bereken_vereenvoudigd(verharding, onderhoud_type: str) -> float:
    # Vereenvoudigd model (9.3.4)
    _, prijs_onderhoud_new = MaintenanceManager.get_maatregel_info(verharding, onderhoud_type)
    _, prijs_lokaal_new = MaintenanceManager.get_maatregel_info(verharding, 'lokaal')
    prijs_onderhoud = prijzen[verharding][onderhoud_type]
    prijs_lokaal = prijzen[verharding]['lokaal']

    T_vereenvoudigd_old = np.sqrt(prijs_onderhoud / prijs_lokaal)

    T_vereenvoudigd = np.sqrt(prijs_onderhoud_new / prijs_lokaal_new)

    return T_vereenvoudigd_old, T_vereenvoudigd
```

# UITGEBREID MODEL

Het laagste CTAT-totaal geeft dus de economisch optimale interventieperiode T voor de onderhouds-maatregel aan.

ook hier gaan we op basis van tabel9 het uitgebreid model berekenen.

- Optimaal tijdstip voor ASFALT met onderhoudstype 'algemeen_v1' = 17
- Optimaal tijdstip voor ASFALT met onderhoudstype 'algemeen_v2' = 21
- Optimaal tijdstip voor ASFALT met onderhoudstype 'versterking_v1' = 26
- Optimaal tijdstip voor ASFALT met onderhoudstype 'versterking_v2' = 28
- Optimaal tijdstip voor ASFALT met onderhoudstype 'versterking_v3' = 30
- __Laatste waarde wijkt wel af van de tabel, 30jaar ipv 33jaar OMDAT WE MAAR VOOR 30 JAAR BEREKENEN__

```python
oppervlakte = 1050
from final_project_stock.ocw_project.ocw_project.config_models import EconomicConfig, EconomicScenario
import json

with open("../ocw_project/_parameters.json", 'r') as f:
    result = json.load(f)
    economic = EconomicConfig(
          standaard=EconomicScenario(**result['economie']['standaard']),
          uitgebreid=EconomicScenario(**result['economie']['uitgebreid'])
    )

print(economic.uitgebreid)

def _bereken_uitgebreid_model(verharding, onderhoud_type: str) -> float:
        """Optimaliseer onderhoudstijdstip volgens de exacte Excel-formules."""
        # Configuratie
        r = economic.uitgebreid.r       # Disconteringsvoet
        i = economic.uitgebreid.i      # Inflatie
        sigma = economic.uitgebreid.sigma # Groeipercentage gewoon onderhoud
        jaren = 30  # Maximale horizon voor analyse

        # Vector van jaren [1, 2, ..., 30]
        t = np.arange(1, jaren + 1)
        
        # Basis kosten (per m² × oppervlakte)
        cr = prijzen[verharding][onderhoud_type] * oppervlakte  # Reparatiekost
        ce = prijzen[verharding]['lokaal'] * oppervlakte * 0.1         # Jaarlijks onderhoud

        # 1. Capital Recovery Factor
        CRF = (r * (1 + r)**t) / ((1 + r)**t - 1)
        
        # 2. Gewoon onderhoud (CE)
        CE = ce * (1 + sigma)**(t-1) * (1 + i)**(t-1)  # Groei met sigma en inflatie
        CEA = CE * (1 + r)**-(t-1)                     # Geactualiseerd
        CEAC = np.cumsum(CEA)                          # Cumulatief geactualiseerd
        CEACE = CRF * CEAC                              # Equivalente annuïteit

        # 3. Reparatiekosten (CR)
        CR = cr * (1 + i)**(t-1)                       # Inflatiecorrectie
        CRA = CR * (1 + r)**-(t-1)                     # Geactualiseerd
        CRACC = CRF * CRA                               # Equivalente annuïteit

        # 4. Totale jaarlijkse kost
        CTA = CEACE + CRACC

        disconteringsvoet, inflatie, sigma_array, reparatie_kost, onderhoud_kost = [
            np.full(30, val) for val in (r, i, sigma, cr, ce)
            ]
        columns = (["CRF", "CE", "CEA", "CEAC", "CEACE",
                     "CR", "CRA", "CRACC", "CTA", "disconteringsvoet",
                    "inflatie", "sigma", "reparatie_kost", "jaarlijks_onderhoud"
                     ])
        data = np.column_stack([CRF, CE, CEA, CEAC, CEACE, CR, 
                                CRA, CRACC, CTA, disconteringsvoet, inflatie, 
                                sigma_array, reparatie_kost, onderhoud_kost])
        new_data= pd.DataFrame(data=data, columns=columns)
        new_data['onderhoud_type'] = onderhoud_type
        
        return np.argmin(CTA) + 1, new_data

def _bereken_uitgebreid_model_new(verharding, onderhoud_type: str) -> float:
        """Optimaliseer onderhoudstijdstip volgens de exacte Excel-formules."""
        # Configuratie
        r = economic.uitgebreid.r       # Disconteringsvoet
        i = economic.uitgebreid.i      # Inflatie
        sigma = economic.uitgebreid.sigma # Groeipercentage gewoon onderhoud
        jaren = 30  # Maximale horizon voor analyse

        # Vector van jaren [1, 2, ..., 30]
        t = np.arange(1, jaren + 1)
        
        # Basis kosten (per m² × oppervlakte)
        _, prijs_onderhoud_new = MaintenanceManager.get_maatregel_info(verharding, onderhoud_type)
        _, prijs_lokaal_new = MaintenanceManager.get_maatregel_info(verharding, 'lokaal')
        cr = prijs_onderhoud_new * oppervlakte      # Reparatiekost
        ce = prijs_lokaal_new * oppervlakte         # Jaarlijks onderhoud

        # 1. Capital Recovery Factor
        CRF = (r * (1 + r)**t) / ((1 + r)**t - 1)
        
        # 2. Gewoon onderhoud (CE)
        CE = ce * (1 + sigma)**(t-1) * (1 + i)**(t-1)  # Groei met sigma en inflatie
        CEA = CE * (1 + r)**-(t-1)                     # Geactualiseerd
        CEAC = np.cumsum(CEA)                          # Cumulatief geactualiseerd
        CEACE = CRF * CEAC                              # Equivalente annuïteit

        # 3. Reparatiekosten (CR)
        CR = cr * (1 + i)**(t-1)                       # Inflatiecorrectie
        CRA = CR * (1 + r)**-(t-1)                     # Geactualiseerd
        CRACC = CRF * CRA                               # Equivalente annuïteit

        # 4. Totale jaarlijkse kost
        CTA = CEACE + CRACC

        disconteringsvoet, inflatie, sigma_array, reparatie_kost, onderhoud_kost = [
            np.full(30, val) for val in (r, i, sigma, cr, ce)
            ]
        columns = (["CRF", "CE", "CEA", "CEAC", "CEACE",
                     "CR", "CRA", "CRACC", "CTA", "disconteringsvoet",
                    "inflatie", "sigma", "reparatie_kost", "jaarlijks_onderhoud"
                     ])
        data = np.column_stack([CRF, CE, CEA, CEAC, CEACE, CR, 
                                CRA, CRACC, CTA, disconteringsvoet, inflatie, 
                                sigma_array, reparatie_kost, onderhoud_kost])
        new_data= pd.DataFrame(data=data, columns=columns)
        new_data['onderhoud_type'] = onderhoud_type
        
        return np.argmin(CTA) + 1, new_data
```

```python
# Tabel 10 – Financieel optimaal tijdstip volgens het niet-vereenvoudigde economische model
jaar_alg1, df_alg1 = _bereken_uitgebreid_model(verharding='asfalt', onderhoud_type='algemeen_v1')
jaar_alg2, df_alg2 = _bereken_uitgebreid_model(verharding='asfalt', onderhoud_type='algemeen_v2')
jaar_alg3, df_alg3 = _bereken_uitgebreid_model(verharding='asfalt', onderhoud_type='algemeen_v3')
jaar_ver1, df_ver1 = _bereken_uitgebreid_model(verharding='asfalt', onderhoud_type='versterking_v1')
jaar_ver2, df_ver2 = _bereken_uitgebreid_model(verharding='asfalt', onderhoud_type='versterking_v2')
jaar_ver3, df_ver3 = _bereken_uitgebreid_model(verharding='asfalt', onderhoud_type='versterking_v3')

jaar_alg1_new, df_alg1_new = _bereken_uitgebreid_model_new(verharding='asfalt', onderhoud_type='algemeen_v1')
jaar_alg2_new, df_alg2_new = _bereken_uitgebreid_model_new(verharding='asfalt', onderhoud_type='algemeen_v2')
jaar_alg3_new, df_alg3_new = _bereken_uitgebreid_model_new(verharding='asfalt', onderhoud_type='algemeen_v3')
jaar_ver1_new, df_ver1_new = _bereken_uitgebreid_model_new(verharding='asfalt', onderhoud_type='versterking_v1')
jaar_ver2_new, df_ver2_new = _bereken_uitgebreid_model_new(verharding='asfalt', onderhoud_type='versterking_v2')
jaar_ver3_new, df_ver3_new = _bereken_uitgebreid_model_new(verharding='asfalt', onderhoud_type='versterking_v3')
```
