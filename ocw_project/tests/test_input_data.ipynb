```python
def classify_road(global_index: float, use_uncertainty_zones: bool = True):
        """Classify road section based on global index with optional uncertainty zones"""
        if use_uncertainty_zones:
            if global_index >= 0.85: return "Zeer goed"
            elif 0.85 > global_index >= 0.75: return "Goed"
            elif 0.75 > global_index >= 0.55: return "Voldoende"
            elif 0.55 > global_index >= 0.45: return "Net voldoende"
            elif 0.45 > global_index >= 0.35: return "Matig"
            elif 0.35 > global_index >= 0.25: return "Onvoldoende"
            else: return "Zeer slecht"
        else:
            if global_index >= 0.80: return (1, "Zeer goed")
            elif 0.80 > global_index >= 0.50: return (2, "Goed")
            elif 0.50 > global_index >= 0.30: return (3, "Voldoende")
            else: return (4, "Onvoldoende")

df_output_uwe['classify'] = df_output_uwe['globaal_index'].apply(lambda x: classify_road(global_index=x, use_uncertainty_zones=True))

df_output_uwe.info()
```

```python
scenario = 18
subset = df_output_uwe[df_output_uwe["scenario_nm"] == scenario]


# Aggregate data by year and quality class
quality_by_year = (subset.groupby(['jaar', 'classify'])
                  .size()
                  .unstack()
                  .fillna(0))

quality_by_year
```
